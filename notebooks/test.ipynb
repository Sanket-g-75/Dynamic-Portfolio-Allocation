## Data Preprocessing

- Remove unnecessary columns and set the index
- Scale all the values (Check if this step can be done in later stage in the project)

```python
import os
import pandas as pd
import numpy as np
import dotenv
import yaml

dotenv.load_dotenv()
```

```python
path = os.environ['PROJECT_FOLDER']
```

```python
with open(os.path.join(path,'config.yaml'),'r') as f:
    config = yaml.safe_load(f)
```

```python
config['features']['commodities']
config['features']['forex']
config['features']['w_indices']
config['features']['other']
```

```python
stocks = []
for i in range(len(config['stocks'])):
    stocks.append(config['stocks'][i][0])
```

```python
features = config['features']['commodities'] + config['features']['forex'] + config['features']['w_indices'] + config['features']['other']
instruments = stocks + features
```

```python
pd.read_csv(os.path.join(path,'data','raw','ADANIENT.NS.csv'))
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
# Stage 1 > Removing unnecessary columns and setting Date as the index

for i in instruments:
    data = pd.read_csv(os.path.join(path,'data','raw',f'{i}.csv'))
    data.drop(columns=[f'{i}_Dividends',f'{i}_StockSplits'],inplace=True)
    data.set_index('Date',inplace=True)

    if i in features:
        data = pd.DataFrame(
            scaler.fit_transform(data),
        columns=data.columns,
        index=data.index
        )

    data.to_csv(os.path.join(path,'data','processed',f'p-{i}.csv'))
```
